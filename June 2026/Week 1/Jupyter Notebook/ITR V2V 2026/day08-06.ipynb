```python
# Day 1 Assignment – Student Grading system

# Inputs
py = float(input("Give me marks of Python subject: "))
db = float(input("Give me marks of DBMS subject: "))
mp = float(input("Give me marks of Microprocessor subject: "))

# Calculate Total & Average Marks
totalMarks = py + db + mp
avgMarks = totalMarks / 3

# Print Total & Average Marks
print(f"Total Marks of Student: {int(totalMarks)}")
print(f"Average Marks of Student: {int(avgMarks)}")

# Assigning Grade
if (True):
    
    if avgMarks >= 90 and avgMarks <= 100:
        print("Grade A")
    elif avgMarks >= 80 and avgMarks < 90:
        print("Grade B")
    elif avgMarks >= 70 and avgMarks < 80:
        print("Grade C")
    elif avgMarks >= 60 and avgMarks < 70:
        print("Grade D")
    elif avgMarks >= 50 and avgMarks < 60:
        print("Grade E")
    elif avgMarks >= 40 and avgMarks < 50:
        print("Grade F")
    else:
        print("Fail")
```
