```python
import pandas as pd
```

```python
data1 = {
    "Name" : ["ABC","DEF","GHI","JKL"],
    "Marks" : [90,67,85,90]
}
df1 = pd.DataFrame(data1)
```

```python
print(df1)
```

```python
print(df1["Marks"]>70)
```

```python
print(df1[df1["Marks"]>70])
```

```python
# sort values function = sort_values()
```

```python
df2 = pd.DataFrame(data1)
df2.sort_values("Marks") # ascending = True
```

```python
df2 = pd.DataFrame(data1)
df2.sort_values("Marks", ascending = False) # descending -> ascending = False
```

```python
# Handling Missing values
```

```python
# Function is used
# df.isnull() 
# df4.dropna()
# df3.fillna(To_fill, inplace = True)
```

```python
import numpy as np
data3 ={
    "Name" : ["ABC","DEF","GHI","JKL"],
    "Marks" : [90,67,np.nan,90]   
}
df3 = pd.DataFrame(data3)
print(df3)
```

```python
# Identifying missing values
df3.isnull()
```

```python
data4 ={
    "Name" : ["ABC","DEF","GHI","JKL"],
    "Marks" : [90,67,np.nan,90]   
}
df4 = pd.DataFrame(data4)
print(df4)
```

```python
# Removing missng value
df4.dropna()
```

```python
# Filling missing values
df3.fillna(0, inplace = True)
```

```python
# Replacing by mean
data5 ={
    "Name" : ["ABC","DEF","GHI","JKL"],
    "Marks" : [90,67,np.nan,90]   
}
df5 = pd.DataFrame(data5)
df5.fillna(df5["Marks"].mean())
```

```python
# Replacing by unknown
data6 ={
    "Name" : ["ABC","DEF","GHI","JKL"],
    "Marks" : [90,67,np.nan,90]   
}
df6 = pd.DataFrame(data6)
df6.fillna("Unknown")
```

```python
# Group by operation
```

```python
emp = {
    "dept" : ["HR","IF","HR","IF"],
    "salary" : [10000,20000,15000,23000]
}
empDF = pd.DataFrame(emp)
```

```python
print(empDF)
```

```python
empDF.groupby("dept")["salary"].mean()
```
